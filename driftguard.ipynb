````python
from torch_geometric.data import Data

def get_normal_subgraph(data, step):
    sub = get_time_step_subgraph(data, step)

    normal_mask = (sub.y == 0)

    # Keep only normal nodes
    x = sub.x[normal_mask]
    y = sub.y[normal_mask]
    time_step = sub.time_step[normal_mask]
    tx_id = sub.tx_id[normal_mask]

    # Remap edges so they point to the new normal-only node indices
    old_to_new = torch.full(
        (sub.num_nodes,),
        -1,
        dtype=torch.long
    )
    old_to_new[normal_mask] = torch.arange(
        normal_mask.sum()
    )

    edge_mask = (
        normal_mask[sub.edge_index[0]]
        & normal_mask[sub.edge_index[1]]
    )

    edge_index = sub.edge_index[:, edge_mask]
    edge_index = old_to_new[edge_index]

    return Data(
        x=x,
        edge_index=edge_index,
        y=y,
        time_step=time_step,
        tx_id=tx_id
    )

print("Function created!")
````
